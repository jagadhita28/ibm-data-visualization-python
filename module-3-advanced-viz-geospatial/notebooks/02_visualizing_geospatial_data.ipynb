# Introduction to Folium <a id="3"></a>


Folium adalah pustaka Python yang andal yang membantu Anda membuat beberapa jenis peta Leaflet. Fakta bahwa hasil Folium bersifat interaktif membuat pustaka ini sangat berguna untuk membangun dasbor.

From the official Folium documentation page:

> Folium memanfaatkan kekuatan pengolahan data dari ekosistem Python dan kekuatan pemetaan dari pustaka Leaflet.js. Manipulasi data Anda di Python, lalu visualisasikan di peta Leaflet melalui Folium.

> Folium memudahkan visualisasi data yang telah dimanipulasi di Python pada peta Leaflet interaktif. Ini memungkinkan pengikatan data ke peta untuk visualisasi kloroplet serta meneruskan visualisasi Vincent/Vega sebagai penanda pada peta.

> Pustaka ini memiliki sejumlah tileset bawaan dari OpenStreetMap, Mapbox, Esri, dan mendukung tileset kustom dengan kunci API Mapbox. Folium mendukung overlay GeoJSON dan TopoJSON, serta pengikatan data ke overlay tersebut untuk membuat peta choropleth dengan skema warna color-brewer.


Membuat peta dunia di **Folium** sangat mudah. Anda cukup membuat objek *Peta* **Folium**, lalu menampilkannya. Yang menarik dari peta **Folium** adalah sifat interaktifnya, sehingga Anda dapat memperbesar area yang diminati terlepas dari tingkat pembesaran awalnya.

# Maps with Markers <a id="4"></a>


Mari kita unduh dan impor data insiden departemen kepolisian menggunakan metode `read_csv()` dari *pandas*.

Unduh dataset dan baca ke dalam dataframe *pandas*:



```python
df_incidents = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Police_Department_Incidents_-_Previous_Year__2016_.csv")

print("Dataset telah diunduh dan dibaca ke dalam dataframe pandas!")
```

Mari kita lihat lima item pertama dalam dataset kita.

```python
df_incidents.head()
```

Jadi setiap baris terdiri dari 13 fitur:
> 1. **IncidntNum**: Nomor Kejadian
> 2. **Category**: Kategori kejahatan atau kejadian
> 3. **Descript**: Deskripsi kejahatan atau kejadian
> 4. **DayOfWeek**: Hari dalam seminggu terjadinya kejadian
> 5. **Date**: Tanggal terjadinya kejadian
> 6. **Time**: Waktu terjadinya kejadian
> 7. **PdDistrict**: Distrik departemen kepolisian
> 8. **Resolution**: Penyelesaian kejahatan dalam hal apakah pelaku ditangkap atau tidak
> 9. **Address**: Alamat terdekat dengan tempat kejadian
> 10. **X**: Nilai bujur lokasi kejahatan
> 11. **Y**: Nilai lintang lokasi kejahatan
> 12. **Location**: Tuple nilai lintang dan bujur
> 13. **PdId**: ID departemen kepolisian

Mari kita cari tahu berapa banyak entri yang ada dalam dataset kita.

```python
df_incidents.shape
```

Jadi, dataframe tersebut terdiri dari 150.500 kejahatan yang terjadi pada tahun 2016. Untuk mengurangi biaya komputasi, mari kita hanya menggunakan 100 insiden pertama dalam dataset ini.

```python
# ambil 100 kejahatan pertama dalam dataframe df_incidents
limit = 100
df_incidents = df_incidents.iloc[0:limit,:]
```

Mari kita pastikan bahwa dataframe kita sekarang hanya berisi 100 kejahatan.

```python
df_incidents.shape
```

Setelah kita sedikit mengurangi data, mari kita visualisasikan di mana kejahatan ini terjadi di kota San Francisco. Kita akan menggunakan gaya default, dan kita akan menginisialisasi tingkat zoom ke 12.

```python
# Nilai lintang dan bujur San Francisco
latitude = 37.77
longitude = -122.42
```

```python
# Buat peta dan tampilkan
sanfran_map = folium.Map(location=[latitude, longitude], zoom_start=12)

# menampilkan peta San Francisco
sanfran_map
```

Sekarang mari kita tumpang tindihkan lokasi kejahatan ke peta. Cara melakukannya di **Folium** adalah dengan membuat *feature* dengan fitur dan gayanya sendiri, lalu menambahkannya ke `sanfran_map`.

```python
# buat instance feature group untuk insiden dalam dataframe
incidents = folium.map.FeatureGroup()

# Lakukan perulangan melalui 100 kejahatan dan tambahkan masing-masing ke feature group incidents
for lat, long in zip(df_incidents.Y, df_incidents.X):
    incidents.add_child(
        folium.vector_layers.CircleMarker(
            [lat, long],
            radius=5, # tentukan ukuran lingkaran penanda yang Anda inginkan
            color="yellow",
            fill=True,
            fill_color="red",
            fill_opacity=0.6
        )
    )

# tambahkan insiden ke peta
sanfran_map.add_child(incidents)
```

Anda juga dapat menambahkan teks pop-up yang akan ditampilkan saat Anda mengarahkan kursor ke sebuah penanda. Mari kita buat setiap penanda menampilkan kategori kejahatan saat kursor diarahkan ke atasnya.

```python
# buat instance feature group incidents dalam dataframe
incidents = folium.map.FeatureGroup()

# lakukan perulangan melalui 100 kejahatan dan tambahkan masing-masing ke feature group incidents
for lat, long in zip(df_incidents.Y, df_incidents.X):
    incidents.add_child(
        folium.vector_layers.CircleMarker(
            [lat, long],
            radius=5, # tentukan ukuran lingkaran penanda yang Anda inginkan
            color="yellow",
            fill=True,
            fill_color="blue",
            fill_opacity=0.6
        )
    )

# tambahkan teks pop-up ke setiap penanda di peta
latitudes = list(df_incidents.Y)
longitudes = list(df_incidents.X)
labels = list(df_incidents.Category)

for lat, long, label in zip(latitudes, longitudes, labels):
    folium.Marker([lat, long], popup=label).add_to(sanfran_map)

# tambahkan insiden ke peta
sanfran_map.add_child(incidents)
```

Bukankah ini sangat keren? Sekarang Anda dapat mengetahui kategori kejahatan apa yang terjadi di setiap penanda.

Jika Anda merasa peta terlalu padat dengan semua penanda ini, ada dua solusi untuk masalah ini. Solusi yang lebih sederhana adalah menghapus penanda lokasi ini dan hanya menambahkan teks ke penanda lingkaran itu sendiri sebagai berikut:

```python
# buat peta dan tampilkan
sanfran_map = folium.Map(location=[latitude, longitude], zoom_start=12)

# lakukan perulangan pada 100 kejahatan dan tambahkan masing-masing ke peta
for lat, long, label in zip(df_incidents.Y, df_incidents.X, df_incidents.Category):
    folium.vector_layers.CircleMarker(
        [lat, long],
        radius=5,
        color="yellow",
        popup=label,
        fill=True,
        fill_color="red",
        fill_opacity=0.6
    ).add_to(sanfran_map)

# tampilkan peta
sanfran_map
```

Solusi tepat lainnya adalah mengelompokkan penanda ke dalam beberapa klaster. Setiap klaster kemudian diwakili oleh jumlah kejahatan di setiap lingkungan. Klaster-klaster ini dapat dianggap sebagai kantong-kantong San Francisco yang kemudian dapat Anda analisis secara terpisah.

Untuk mengimplementasikannya, kita mulai dengan membuat objek *MarkerCluster* dan menambahkan semua titik data dalam dataframe ke objek ini

```python
from folium import plugins
```

```python
# mari kita mulai lagi dengan salinan peta San Francisco yang bersih.
sanfran_map = folium.Map(location=[latitude, longitude], zoom_start=12)

# buat instance objek marker cluster untuk incidents (insiden) dalam dataframe
incidents = plugins.MarkerCluster().add_to(sanfran_map)

# lakukan perulangan melalui dataframe dan tambahkan setiap titik data ke marker cluster
for lat, long, label in zip(df_incidents.Y, df_incidents.X, df_incidents.Category):
    folium.Marker(
        location=[lat, long],
        icon=None,
        popup=label
    ).add_to(incidents)

# tampilkan peta
sanfran_map
```

Perhatikan bagaimana ketika Anda memperkecil tampilan sepenuhnya, semua penanda dikelompokkan menjadi satu kluster, *kluster global*, yang terdiri dari 100 penanda atau kejahatan, yang merupakan jumlah total kejahatan dalam dataframe kita. Setelah Anda mulai memperbesar tampilan, *kluster global* akan mulai terpecah menjadi kluster yang lebih kecil. Memperbesar tampilan sepenuhnya akan menghasilkan penanda individual.
