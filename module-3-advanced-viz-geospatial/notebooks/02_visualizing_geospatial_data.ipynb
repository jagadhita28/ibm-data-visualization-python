# Introduction to Folium <a id="3"></a>


Folium adalah pustaka Python yang andal yang membantu Anda membuat beberapa jenis peta Leaflet. Fakta bahwa hasil Folium bersifat interaktif membuat pustaka ini sangat berguna untuk membangun dasbor.

From the official Folium documentation page:

> Folium memanfaatkan kekuatan pengolahan data dari ekosistem Python dan kekuatan pemetaan dari pustaka Leaflet.js. Manipulasi data Anda di Python, lalu visualisasikan di peta Leaflet melalui Folium.

> Folium memudahkan visualisasi data yang telah dimanipulasi di Python pada peta Leaflet interaktif. Ini memungkinkan pengikatan data ke peta untuk visualisasi kloroplet serta meneruskan visualisasi Vincent/Vega sebagai penanda pada peta.

> Pustaka ini memiliki sejumlah tileset bawaan dari OpenStreetMap, Mapbox, Esri, dan mendukung tileset kustom dengan kunci API Mapbox. Folium mendukung overlay GeoJSON dan TopoJSON, serta pengikatan data ke overlay tersebut untuk membuat peta choropleth dengan skema warna color-brewer.


Membuat peta dunia di **Folium** sangat mudah. Anda cukup membuat objek *Peta* **Folium**, lalu menampilkannya. Yang menarik dari peta **Folium** adalah sifat interaktifnya, sehingga Anda dapat memperbesar area yang diminati terlepas dari tingkat pembesaran awalnya.


```python
# mendefinisikan peta dunia
world_map = folium.Map()

# menampilkan peta dunia
world_map
```

Silakan. Coba perbesar dan perkecil peta yang ditampilkan di atas.

Anda dapat menyesuaikan definisi default peta dunia ini dengan menentukan pusat peta Anda, dan tingkat zoom awal.

Semua lokasi pada peta ditentukan oleh nilai *Lintang* dan *Bujur* masing-masing. Jadi Anda dapat membuat peta dan memasukkan pusat dengan nilai *Lintang* dan *Bujur* **[0, 0]**.

Untuk pusat yang telah ditentukan, Anda juga dapat menentukan tingkat zoom awal ke lokasi tersebut saat peta ditampilkan. **Semakin tinggi tingkat zoom, semakin peta diperbesar ke pusat**.

Mari kita buat peta yang berpusat di sekitar Kanada dan bereksperimen dengan tingkat zoom untuk melihat bagaimana pengaruhnya terhadap peta yang ditampilkan.

```python
# tentukan peta dunia yang berpusat di sekitar Kanada dengan tingkat zoom rendah
canada_map = folium.Map(location=[57.45323300961382, -101.66674931624746], zoom_start=4)

# menampilkan peta canada
canada_map
```

Mari kita buat peta lagi dengan tingkat zoom yang lebih tinggi.

```python
# tentukan peta dunia yang berpusat di sekitar Kanada dengan tingkat zoom yang lebih tinggi
canada_map = folium.Map(location=[57.45323300961382, -101.66674931624746], zoom_start=8)

# menampilkan peta canada
canada_map
```

Seperti yang Anda lihat, semakin tinggi tingkat zoom, semakin peta diperbesar ke titik tengah yang diberikan.

### A. Dark Gray Canvas Maps (Peta Kanvas Abu-abu Gelap)

Ini adalah peta hitam putih (B+W) dengan kontras tinggi. Peta ini sangat cocok untuk penggabungan data dan menjelajahi kelokan sungai serta zona pesisir.

Mari kita buat peta Kanada dengan kanvas abu-abu gelap dan tingkat zoom 4.

```python
# buat peta dunia Esri Dark Gray Canvas yang berpusat di Kanada
canada_map = folium.Map(location=[57.45323300961382, -101.66674931624746], zoom_start=4, max_zoom=16,
                       tiles="https://services.arcgisonline.com/ArcGIS/rest/services/Canvas/World_Dark_Gray_Base/MapServer/tile/{z}/{y}/{x}",
                       attr="Esri, HERE, Garmin, © OpenStreetMap contributors")

# menampilkan peta canada
canada_map
```

Silakan perbesar dan perkecil tampilan untuk melihat bagaimana gaya ini dibandingkan dengan gaya bawaan.

### B. Light Gray Canvas Maps (Peta Kanvas Abu-abu Muda)

Peta Kanvas Abu-abu Muda dirancang dengan estetika yang ringan dan minimalis. Peta ini memiliki latar belakang putih atau berwarna terang dan menampilkan garis-garis sederhana dan bersih untuk elemen peta. Peta ini dikenal karena desainnya yang modern dan menarik secara visual.

Mari kita buat peta Kanada dengan kanvas abu-abu muda dan tingkat zoom 4.

```python
# buat peta dunia Esri Light Gray Canvas yang berpusat di Kanada
canada_map = folium.Map(location=[57.45323300961382, -101.66674931624746], zoom_start=4,
                       tiles="https://services.arcgisonline.com/ArcGIS/rest/services/Canvas/World_Light_Gray_Base/MapServer/tile/{z}/{y}/{x}",
                       attr="Esri, HERE, Garmin, © OpenStreetMap contributors", max_zoom=16)

# tampilkan peta
canada_map
```

Silakan perbesar dan perkecil tampilan untuk melihat perbandingan gaya ini dengan Dark Gray Canvas dan gaya default.

Perbesar tampilan dan perhatikan bagaimana perbatasan mulai terlihat saat Anda memperbesar, dan nama-nama negara yang ditampilkan dalam bahasa Inggris.
