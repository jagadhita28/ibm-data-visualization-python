## *pandas* Intermediate: Indexing and Selection (slicing)<a id="3"></a>


### Select Column
**Ada dua cara untuk memfilter berdasarkan nama kolom:**

Method 1: Cepat dan mudah, tetapi hanya berfungsi jika nama kolom TIDAK mengandung spasi atau karakter khusus.
```python
    df.column_name               # returns series
```

Method 2: Lebih tangguh, dan dapat melakukan penyaringan pada beberapa kolom.

```python
    df['column']                  # returns series
```

```python 
    df[['column 1', 'column 2']]  # returns dataframe
```
---

### *pandas* Filtering based on a criteria <a id="4"></a>
Untuk memfilter dataframe berdasarkan suatu kondisi, cukup memberikan kondisi tersebut sebagai vektor boolean.

Sebagai contoh, filter dataframe untuk menampilkan data tentang negara-negara Asia (AreaName = Asia).

````python
# 1. membuat rangkaian condition boolean
condition = df_can["Continent"] == "Asia"
print(condition)
````

````python
# 2. meneruskan kondisi ini ke dataframe
df_can[condition]
````

````python
# dalam satu baris dapat memasukkan beberapa kriteria.
# sekarang lakukan filter untuk AreaNAME = Asia dan RegName = Southern Asia

df_can[(df_can["Continent"] == "Asia") & (df_can["Region"] == "Southern Asia")]
````

**Catatan**: Saat menggunakan operator 'and' dan 'or', *pandas* mengharuskan untuk menggunakan '&' dan '|' sebagai pengganti 'and' dan 'or'.

Jangan lupa untuk mengapit kedua kondisi dalam tanda kurung

**Latihan:** Ambil data di mana AreaName adalah 'Africa' dan RegName adalah 'Southern Africa'. <br>Tampilkan dataframe dan cari tahu berapa banyak instance yang ada?

````python
south_africa = (df_can["Continent"] == "Africa") & (df_can["Region"] == "Southern Africa")
df_can[south_africa]
````

### *pandas* Sorting Values of a Dataframe or Series <a id="5"></a><br>

Gunakan fungsi `sort_values()` untuk mengurutkan DataFrame atau Series berdasarkan satu atau lebih kolom. <br>Tentukan terlebih dahulu kolom yang ingin digunakan untuk mengurutkan dan urutannya (naik atau turun). Berikut sintaks penggunaannya:-<br><br>
```df.sort_values(col_name, axis=0, ascending=True, inplace=False, ignore_index=False)```<br><br>
**col_name** - kolom yang akan diurutkan. <br>
**axis** - sumbu yang akan digunakan untuk mengurutkan. 0 untuk pengurutan berdasarkan baris (default) dan 1 untuk pengurutan berdasarkan kolom.<br>
**ascending** - untuk mengurutkan dalam urutan menaik (True, default) atau urutan menurun (False).<br>
**inplace** - untuk melakukan operasi pengurutan di tempat (True) atau mengembalikan salinan yang sudah diurutkan (False, default).<br>
**ignore_index** - untuk mengatur ulang indeks setelah pengurutan (True) atau mempertahankan nilai indeks asli (False, default).<br>


Urutkan dataframe df_can berdasarkan kolom 'Total', dalam urutan menurun untuk menemukan 5 negara teratas yang paling banyak berkontribusi terhadap imigrasi ke Kanada.

````python
df_can.sort_values("Total", axis=0, ascending=False, inplace=True)

top_5 = df_can.head(5)
top_5
````

**Latihan:** Temukan 3 negara teratas yang paling banyak menyumbang imigrasi ke Kanada pada tahun 2010. <br> Tampilkan nama negara beserta jumlah imigran pada tahun tersebut.

````python
df_can.sort_values(["2010"], axis=0, ascending=False, inplace=True)

top_3_2010 = df_can[["Continent","Region","2010"]].head(3)
top_3_2010
````
