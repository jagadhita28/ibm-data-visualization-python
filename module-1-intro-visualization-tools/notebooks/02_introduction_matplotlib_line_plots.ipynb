# Line Pots (Series/Dataframe)<a id="4"></a>


**Apa itu grafik garis dan mengapa menggunakannya?**

Grafik garis atau plot garis adalah jenis grafik yang menampilkan informasi sebagai serangkaian titik data yang disebut 'markers' yang dihubungkan oleh segmen garis lurus. Ini adalah jenis grafik dasar yang umum di banyak bidang.
Gunakan grafik garis ketika Anda memiliki kumpulan data kontinu. Grafik ini paling cocok untuk visualisasi data berbasis tren selama periode waktu tertentu.

**Mari kita mulai dengan studi kasus:**

Pada tahun 2010, Haiti mengalami gempa bumi dahsyat berkekuatan 7,0 skala Richter. Gempa tersebut menyebabkan kerusakan dan kehilangan nyawa yang meluas, dan sekitar tiga juta orang terkena dampak bencana alam ini. Sebagai bagian dari upaya kemanusiaan Kanada, Pemerintah Kanada meningkatkan upayanya dalam menerima pengungsi dari Haiti. Kita dapat dengan cepat memvisualisasikan upaya ini menggunakan plot `Line`:

**Pertanyaan:** Gambarkan grafik garis imigrasi dari Haiti menggunakan `df.plot()`.

Pertama, kita akan mengekstrak rangkaian data untuk Haiti.

````python
# Buat data series
haiti = df_can.loc["Haiti", years] # beralih pada tahun 1980 - 2013 untuk mengecualikan kolom 'total'.
haiti.head()
````

Selanjutnya, kita akan membuat plot garis dengan menambahkan `.plot()` ke dataframe `haiti`.

````python
haiti.plot()
````

*pandas* secara otomatis mengisi sumbu x dengan nilai indeks (tahun), dan sumbu y dengan nilai kolom (populasi). <br>

Selain itu, mari kita beri label pada sumbu x dan y menggunakan `plt.title()`, `plt.ylabel()`, dan `plt.xlabel()` sebagai berikut:

````python
haiti.plot(kind="line", figsize=(10, 4))

plt.ylabel("Jumlah Imigran")
plt.xlabel("Tahun")
plt.title("Imigran dari Haiti")

plt.savefig("../outputs/imigran_dari_haiti.png", dpi=200,
           bbox_inches="tight")
plt.show()
````

Kita dapat dengan jelas melihat bagaimana jumlah imigran dari Haiti melonjak sejak tahun 2010 seiring dengan meningkatnya upaya Kanada untuk menerima pengungsi dari Haiti. Mari kita beri anotasi pada lonjakan ini dalam plot dengan menggunakan metode `plt.text()`.<br>
<br>
Namun, perhatikan bahwa tahun bertipe *string*. Mari kita ubah tipe nilai indeks menjadi *integer* terlebih dahulu.

````python
haiti.index = haiti.index.map(int)
````

````python
haiti.plot(kind="line", figsize=(10, 4))

plt.title("Imigran dari Haiti")
plt.ylabel("Jumlah Imigran")
plt.xlabel("Tahun")

# Berikan anotasi pada Gempa Bumi tahun 2010.
plt.text(2004, 6000, "Gempa Bumi 2010")

plt.savefig("../outputs/imigran_dari_haiti_anotasi.png", dpi=200,
           bbox_inches="tight")
plt.show()
````

Hanya dengan beberapa baris kode, Anda dapat dengan cepat mengidentifikasi dan memvisualisasikan lonjakan imigrasi!

Catatan singkat tentang nilai x dan y dalam `plt.text(x, y, label)`:

Karena sumbu x (tahun) bertipe 'integer', kita menentukan x sebagai tahun. Sumbu y (jumlah imigran) bertipe 'integer', jadi kita cukup menentukan nilai y = 6000.

```python
plt.text(2000, 6000, 'Gempa Bumi 2010') # tahun disimpan sebagai tipe int
```
Jika tahun disimpan sebagai tipe 'string', kita perlu menentukan x sebagai posisi indeks tahun tersebut. Misalnya, indeks ke-20 adalah tahun 2000 karena merupakan tahun ke-20 dengan tahun dasar 1980.
```python
plt.text(20, 6000, 'Gempa Bumi 2010') # tahun disimpan sebagai tipe int
```
Kita akan membahas metode anotasi tingkat lanjut di modul selanjutnya.

Kita dapat dengan mudah menambahkan lebih banyak negara ke dalam grafik garis untuk membuat perbandingan yang bermakna mengenai imigrasi dari berbagai negara.

**Pertanyaan:** Mari kita bandingkan jumlah imigran dari India dan Tiongkok dari tahun 1980 hingga 2013.

Langkah 1: Dapatkan kumpulan data untuk China dan India, lalu tampilkan dataframe tersebut.

````python
df_CI = df_can.loc[["India", "China"], years]
df_CI.head()
````

Langkah 2: Menggambar grafik. Kita akan secara eksplisit menentukan plot garis dengan memberikan parameter `kind` ke `plot()`.

````python
df_CI.plot(kind="line", figsize=(10, 4))
````

Plotnya tidak terlihat benar...

Ingat bahwa *pandas* memplot indeks pada sumbu x dan kolom sebagai garis individual pada sumbu y. Karena `df_CI` adalah dataframe dengan `country` sebagai indeks dan `years` sebagai kolom, kita harus terlebih dahulu mentranspos dataframe menggunakan metode `transpose()` untuk menukar baris dan kolom.

````python
df_CI = df_CI.transpose()
df_CI.head()
````

*pandas* akan secara otomatis membuat grafik kedua negara tersebut pada grafik yang sama. Silakan buat plot dataframe yang telah ditransposkan. Pastikan untuk menambahkan judul pada plot dan memberi label pada sumbu.

````python
# Ubah indeks menjadi integer
df_CI.index = df_CI.index.map(int)
df_CI.plot(kind="line", figsize=(10, 4))

plt.title("Imigran dari India dan China")
plt.ylabel("Jumlah Imigran")
plt.xlabel("Tahun")

plt.savefig("../outputs/imigran_dari_india_dan_china", dpi=200,
           bbox_inches="tight")
plt.show()
````

<br>Dari grafik di atas, kita dapat mengamati bahwa Tiongkok dan India memiliki tren imigrasi yang sangat mirip selama bertahun-tahun.

*Catatan*: Mengapa kita tidak perlu mentranspos dataframe Haiti sebelum membuat plot (seperti yang kita lakukan untuk df_CI)?

Itu karena `haiti` adalah sebuah series, bukan dataframe, dan memiliki tahun sebagai indeksnya seperti yang ditunjukkan di bawah ini.


Line plot adalah alat yang berguna untuk menampilkan beberapa variabel dependen terhadap satu variabel independen. Namun, disarankan agar tidak lebih dari 5-10 garis pada satu grafik; jika lebih dari itu, akan sulit untuk diinterpretasikan

**Pertanyaan:** Bandingkan tren 5 negara teratas yang paling banyak berkontribusi terhadap imigrasi ke Kanada.

````python
# Urutkan jumlah imigrasi dari total tertinggi ke terendah
df_can.sort_values("Total", axis=0, ascending=False,
                  inplace=True)
````

````python
# Buat object dengan mengambil 5 negara teratas
top_5 = df_can.head()
top_5
````

````python
# Transpose dataframenya
top_5 = top_5[years].transpose()
top_5.head()
````

````python
top_5.index
````

````python
# Buat line plot untuk melihat perbandingan dari 5 negara tersebut
top_5.index = top_5.index.map(int)
top_5.plot(kind="line", figsize=(14, 8))

plt.title("Trend Imigran dari 5 Negara Teratas")
plt.ylabel("Jumlah Imigran")
plt.xlabel("Tahun")

plt.savefig("../outputs/trend_imigran_5_negara_teratas", dpi=200,
           bbox_inches="tight")
plt.show()
````
